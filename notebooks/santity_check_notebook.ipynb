### Basic Set up

```python
import gymnasium as gym
import rware

# The environment name format is rware-{size}-{agents}ag-v2
ENV_STRING =  "rware-tiny-1ag-v2"
env = gym.make(ENV_STRING, render_mode="rgb_array")

class ExtendEnv(gym.Wrapper):
    def __init__(self, env):
        super().__init__(env)
    def render(self):
        frame = self.env.unwrapped.render(mode="rgb_array")
        return frame
```

### Action space

A=`{ Turn Left, Turn Right, Forward, Load/Unload Shelf }`

```python
env.action_space
```

### Observation Space

`Note Game is partially observable for each agent`

1. `3x3`grid size per observation per agent
2. information that `3x3` grid array contains
    - location, rotation, load status of `agent`
    - location, rotation of `other agents`
    - Shelves and status of Shelves e.g. whether this shelve is target shelve to drop boxes at or not

`so per single agent single obs is of (71,) shape, flatten 3x3 grid into (71,)`

```python
from gymnasium.wrappers import RecordVideo, RecordEpisodeStatistics

VIDEO_FOLDER="video_folder"
NUM_EVAL_EPISODES=10
ENV_STRING =  "rware-tiny-1ag-v2"
env = gym.make(ENV_STRING, render_mode="rgb_array")
env = ExtendEnv(env)
env = RecordVideo(env , name_prefix="eval", video_folder=VIDEO_FOLDER)
env = RecordEpisodeStatistics(env, buffer_length=NUM_EVAL_EPISODES)


def evaluate_agent():
    try:
        for ep in range(NUM_EVAL_EPISODES):
            done = False
            episode_reward = 0
            step_count = 0
            obs, info = env.reset()
            print("="*10)
            print(f"Episode {ep} Started!")
            while not done:
                action = env.action_space.sample()
                obs, reward, terminated, truncated, info = env.step(action)
                episode_reward += reward
                step_count+= 1
                done = terminated or truncated
            print(f"Episode {ep} Ended!")
            print(f"Step Count is {step_count}")
            print(f"Episode Reward is {episode_reward}")
            print("="*10)

    except Exception as e:
        print(f"error occurred {e}")
    finally:
        env.close()
```

```python
evaluate_agent()
```
